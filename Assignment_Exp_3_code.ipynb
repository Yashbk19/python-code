```
book = {'title':'Python','author':'ABC','price':299,'pages':350}
book['publisher'] = 'PubHouse'
book['price'] = 249
del book['pages']
print('author' in book)
print(book)
# Shows before/after operations
```

```
emp = {'id':59,'name':'Yash','department':'IT','salary':50000}
emp['bonus'] = 5000
emp['salary'] = 55000
del emp['department']
print(emp)
emp.clear()
print(emp)
# Employee dict operations
```

```
fruits = {'apple':50,'banana':30,'orange':40,'grapes':120}
fruits['mango'] = 80
fruits['apple'] = 55
del fruits['banana']
popped = fruits.pop('orange')
print(fruits.keys(), fruits.values(), fruits.items())
# Fruit dict operations
```

```
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print('union', s1 | s2)
print('intersection', s1 & s2)
print('difference', s1 - s2)
print('sym diff', s1 ^ s2)
s1.add(7)
s1.remove(1)
print(2 in s1)
print(s1.issubset({1,2,3,4,7}))
print(s1.issuperset({2,3}))
# Set operations
```

```
c1 = {'red','blue','green'}
c2 = {'yellow','blue','pink'}
c1.add('black')
c2.add('white')
c1.remove('green')
print(c1 | c2)
print(c1 & c2)
print(c1 - c2)
print(c1 ^ c2)
print(c1.issubset(c2))
print(c1.issuperset(c2))
# Color set operations
```

```
nums = set(range(1,11))
nums.update([11,12,13])
nums.discard(2)
nums.remove(3)
print(nums)
# union etc can be applied with another set
```

```
capitals = {'India':'New Delhi','USA':'Washington','Japan':'Tokyo','UK':'London'}
capitals['France'] = 'Paris'
capitals['Germany'] = 'Berlin'
capitals['India'] = 'New Delhi Updated'
del capitals['UK']
popped = capitals.pop('Japan')
print(capitals.keys(), capitals.values(), capitals.items())
print('India' in capitals)
# Capitals dict ops
```

```
movies = {'Inception','Avatar','Interstellar','Gladiator','Titanic','TheGodfather','Joker','Matrix'}
movies.update(['3 Idiots','PK'])
movies.discard('Titanic')
print(movies | {'SomeMovie'})
print(movies & {'Inception','Joker'})
print(movies - {'PK'})
print(movies ^ {'OldMovie','PK'})
print(movies.issubset({'Inception','Avatar','Interstellar','Gladiator','Titanic','TheGodfather','Joker','Matrix','3 Idiots','PK'}))
# Movies set ops
```
