```
n = -5
if n > 0:
    print('Positive')
elif n == 0:
    print('Zero')
else:
    print('Negative')
# Example: -5 -> Negative
```

```
n = 6
if n % 2 == 0:
    print('Even')
else:
    print('Odd')
# Example: 6 -> Even
```

```
a, b, c = 10, 20, 15
if a >= b and a >= c:
    print('Largest is', a)
elif b >= a and b >= c:
    print('Largest is', b)
else:
    print('Largest is', c)
# Output: Largest is 20
```

```
marks = 85
if marks >= 90:
    print('A')
elif marks >= 75:
    print('B')
elif marks >= 50:
    print('C')
else:
    print('F')
# Example: 85 -> B
```

```
year = 2024
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print('Leap year')
else:
    print('Not leap year')
# 2024 -> Leap year
```

```
age = 25
if age < 13:
    print('Child')
elif age < 20:
    print('Teenager')
elif age < 60:
    print('Adult')
else:
    print('Senior')
# Example: 25 -> Adult
```

```
units = 250
if units <= 100:
    bill = units * 1
elif units <= 200:
    bill = 100*1 + (units-100)*2
else:
    bill = 100*1 + 100*2 + (units-200)*3
print('Bill =', bill)
# Example calculation
```

```
a, b, c = 5,5,5
if a == b == c:
    print('Equilateral')
elif a == b or b == c or a == c:
    print('Isosceles')
else:
    print('Scalene')
# Example: equilateral
```

```
day_map = {1:'Monday',2:'Tuesday',3:'Wednesday',4:'Thursday',5:'Friday',6:'Saturday',7:'Sunday'}
print(day_map.get(3, 'Invalid'))
# 3 -> Wednesday
```

```
ops = {'+': lambda x,y: x+y, '-': lambda x,y: x-y, '*': lambda x,y: x*y, '/': lambda x,y: x/y}
print(ops['+'](10,5))
# 10+5 -> 15
```

```
ch = 'a'
if ch.lower() in 'aeiou':
    print('Vowel')
else:
    print('Consonant')
# a -> Vowel
```

```
n = 30
if n % 2 == 0:
    print('Divisible by 2')
if n % 3 == 0:
    print('Divisible by 3')
if n % 5 == 0:
    print('Divisible by 5')
# 30 -> divisible by all
```
