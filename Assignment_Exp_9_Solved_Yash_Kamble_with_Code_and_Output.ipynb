```python
class Student:
    school = "SGU"
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):
        print(f"Name: {self.name}, Marks: {self.marks}, School: {Student.school}")
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school
    @staticmethod
    def grade(marks):
        return "A" if marks >= 75 else "B" if marks >= 50 else "C"
    def __del__(self):
        print(f"Student {self.name} record deleted")

s = Student("Amit", 82)
s.display()
print("Grade:", Student.grade(s.marks))
Student.change_school("New SGU")
s2 = Student("Rita", 48)
s2.display()
```

```python
class Employee:
    company = "ABC Pvt Ltd"
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary
    def yearly_salary(self):
        return self.salary * 12
    @classmethod
    def update_company(cls, new_company):
        cls.company = new_company
    @staticmethod
    def bonus(salary):
        return salary * 0.10
    def __del__(self):
        print(f"Employee {self.name} record removed")

e = Employee(101, "Sunil", 40000)
print("Yearly Salary:", e.yearly_salary())
print("Bonus:", Employee.bonus(e.salary))
Employee.update_company("XYZ Pvt Ltd")
print("Company updated to:", Employee.company)
```

```python
class BankAccount:
    branch = "Main"
    def __init__(self, acc_no, balance=0):
        self.acc_no = acc_no
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
    @classmethod
    def update_branch(cls, new_branch):
        cls.branch = new_branch
    @staticmethod
    def interest(balance):
        return balance * 0.05
    def __del__(self):
        print("Account closed")

a = BankAccount("AC123", 1000)
a.deposit(500)
print("Balance after deposit:", a.balance)
a.withdraw(300)
print("Balance after withdrawal:", a.balance)
print("Interest on balance:", BankAccount.interest(a.balance))
```

```python
class Rectangle:
    unit = "cm"
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
    @classmethod
    def update_unit(cls, new_unit):
        cls.unit = new_unit
    @staticmethod
    def compare(r1, r2):
        return r1.area() - r2.area()
    def __del__(self):
        print("Rectangle deleted")

r1 = Rectangle(4,5)
r2 = Rectangle(3,8)
print("Area r1:", r1.area())
print("Perimeter r1:", r1.perimeter())
print("Area difference (r1 - r2):", Rectangle.compare(r1,r2))
```

```python
class Car:
    total_cars = 0
    def __init__(self, brand, model, price):
        Car.total_cars += 1
        self.brand = brand
        self.model = model
        self.price = price
    def display(self):
        print(f"{self.brand} {self.model} - ₹{self.price}")
    @classmethod
    def update_total(cls, count):
        cls.total_cars = count
    @staticmethod
    def discount(price, percent):
        return price - (price * percent / 100)
    def __del__(self):
        print("Car removed")

c = Car("Hyundai","i20",800000)
c.display()
print("Discounted price (10%):", Car.discount(c.price,10))
print("Total cars:", Car.total_cars)
```

```python
import math
class Circle:
    pi = math.pi
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return Circle.pi * self.radius * self.radius
    def circumference(self):
        return 2 * Circle.pi * self.radius
    @classmethod
    def update_pi(cls, new_pi):
        cls.pi = new_pi
    @staticmethod
    def validate_radius(r):
        return r > 0
    def __del__(self):
        print("Circle removed")

c = Circle(5)
print("Area:", round(c.area(),2))
print("Circumference:", round(c.circumference(),2))
print("Valid radius?:", Circle.validate_radius(c.radius))
```

```python
class Book:
    publisher = "DefaultPub"
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    def display(self):
        print(f"{self.title} by {self.author} - ₹{self.price}")
    @classmethod
    def update_publisher(cls, new_pub):
        cls.publisher = new_pub
    @staticmethod
    def price_limit(price):
        return price <= 500
    def __del__(self):
        print("Book record removed")

b = Book("Python Basics","Kamal",450)
b.display()
print("Under price limit?:", Book.price_limit(b.price))
Book.update_publisher("SGU Press")
print("Publisher:", Book.publisher)
```

```python
class Laptop:
    warranty_years = 1
    def __init__(self, brand, model, ram):
        self.brand = brand
        self.model = model
        self.ram = ram
    def display(self):
        print(f"{self.brand} {self.model}, RAM: {self.ram}GB")
    @classmethod
    def update_warranty(cls, years):
        cls.warranty_years = years
    @staticmethod
    def sufficient_ram(ram):
        return ram >= 8
    def __del__(self):
        print("Laptop removed")

l = Laptop("Dell","Inspiron",16)
l.display()
print("Sufficient RAM?:", Laptop.sufficient_ram(l.ram))
Laptop.update_warranty(2)
print("Warranty (yrs):", Laptop.warranty_years)
```

```python
class Payroll:
    hra_percent = 10
    def __init__(self, name, basic_salary):
        self.name = name
        self.basic = basic_salary
    def total_salary(self):
        hra = self.basic * Payroll.hra_percent/100
        return self.basic + hra
    @classmethod
    def update_hra(cls, percent):
        cls.hra_percent = percent
    @staticmethod
    def tax(amount):
        return amount * 0.05
    def __del__(self):
        print("Payroll removed")

p = Payroll("Ramesh", 50000)
print("Total salary:", p.total_salary())
print("Tax on basic:", Payroll.tax(p.basic))
Payroll.update_hra(12)
print("Updated HRA%:", Payroll.hra_percent)
```

```python
class Temperature:
    boiling_point_c = 100.0
    def __init__(self, celsius):
        self.c = celsius
    def to_fahrenheit(self):
        return (self.c * 9/5) + 32
    @classmethod
    def set_boiling_point(cls, bp):
        cls.boiling_point_c = bp
    @staticmethod
    def validate_temp(t):
        return -273.15 <= t
    def __del__(self):
        print("Temperature object deleted")

t = Temperature(25)
print("Fahrenheit:", round(t.to_fahrenheit(),2))
print("Is valid temp?:", Temperature.validate_temp(t.c))
```

```python
class Product:
    tax_rate = 0.05
    def __init__(self, name, price, qty):
        self.name = name
        self.price = price
        self.qty = qty
    def total_cost(self):
        return self.price * self.qty * (1 + Product.tax_rate)
    @classmethod
    def update_tax(cls, rate):
        cls.tax_rate = rate
    @staticmethod
    def validate_qty(q):
        return q >= 0
    def __del__(self):
        print("Product removed")

pr = Product("Pen", 10, 5)
print("Total cost:", pr.total_cost())
print("Valid qty?:", Product.validate_qty(pr.qty))
```

```python
class Attendance:
    total_working_days = 30
    def __init__(self, emp_name, days_present):
        self.emp_name = emp_name
        self.days_present = days_present
    def display(self):
        print(f"{self.emp_name}: {self.days_present}/{Attendance.total_working_days}")
    @classmethod
    def update_total_days(cls, days):
        cls.total_working_days = days
    @staticmethod
    def bonus_eligible(days_present):
        return days_present >= 28
    def __del__(self):
        print("Attendance record removed")

a = Attendance("Kiran", 29)
a.display()
print("Bonus eligible?:", Attendance.bonus_eligible(a.days_present))
```

```python
import math
class CircleMath:
    pi = math.pi
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return CircleMath.pi * self.radius**2
    def circumference(self):
        return 2 * CircleMath.pi * self.radius
    @classmethod
    def update_pi(cls, newpi):
        cls.pi = newpi
    @staticmethod
    def compare_area(c1, c2):
        return c1.area() - c2.area()
    def __del__(self):
        print("CircleMath removed")

c1 = CircleMath(3)
c2 = CircleMath(4)
print("Area c1:", round(c1.area(),2))
print("Area c2:", round(c2.area(),2))
print("Area difference:", round(CircleMath.compare_area(c1,c2),2))
```

```python
class StudentMarks:
    passing_marks = 40
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks  # list
    def total(self):
        return sum(self.marks)
    def average(self):
        return self.total()/len(self.marks)
    @classmethod
    def update_passing(cls, marks):
        cls.passing_marks = marks
    @staticmethod
    def assign_grade(avg):
        if avg >= 75: return "A"
        if avg >= 50: return "B"
        return "C"
    def __del__(self):
        print("StudentMarks removed")

s = StudentMarks("Tina", [80,70,90])
print("Total:", s.total())
print("Average:", s.average())
print("Grade:", StudentMarks.assign_grade(s.average()))
```

```python
class Vehicle:
    total_count = 0
    def __init__(self, vtype, brand, speed):
        Vehicle.total_count += 1
        self.vtype = vtype
        self.brand = brand
        self.speed = speed
    def display(self):
        print(f"{self.vtype} {self.brand} speed {self.speed} km/h")
    @classmethod
    def update_total(cls, cnt):
        cls.total_count = cnt
    @staticmethod
    def legal_speed(speed):
        return speed <= 80
    def __del__(self):
        print("Vehicle removed")

v = Vehicle("Bike","Hero",75)
v.display()
print("Legal speed?:", Vehicle.legal_speed(v.speed))
print("Total vehicles:", Vehicle.total_count)
```
