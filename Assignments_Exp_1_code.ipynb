```
name = 'Yash'
age = 21
height = 166.5
student = True
print(name, type(name))
print(age, type(age))
print(height, type(height))
print(student, type(student))
# Output: Yash <class 'str'> 21 <class 'int'> 166.5 <class 'float'> True <class 'bool'>
```

```
a = int(12)
b = int(5)
print('Sum :', a + b)
print('Diff :', a - b)
print('Div :', a / b)
print('Multi :', a * b)
print('Mod :', a % b)
print('Expo :', a ** b)
print('floor div :', a // b)
# Sample Input: 12, 5
```

```
a = 12
b = 5
print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
# Outputs are booleans
```

```
a = True
b = False
print('AND :', a and b)
print('OR :', a or b)
print('NOT b :', not b)
print('NOT a :', not a)
# Output: AND : False, OR : True, NOT b : True, NOT a : False
```
