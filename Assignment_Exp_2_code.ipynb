```
lst = [0,1,2,2,3,4,5,6,6,8]
lst.sort()
print(lst)
# Output: sorted ascending
```

```
lst = [0,1,2,2,3,4,5,6,6,8]
lst.sort(reverse=True)
print(lst)
# Output: sorted descending
```

```
t = (6,3,1,5,2,4)
sorted_t = tuple(sorted(t))
print(sorted_t)
# Output: sorted tuple
```

```
t1 = (1,2,3)
t2 = (4,5)
merged = t1 + t2
print(merged)
# Output: (1,2,3,4,5)
```

```
lst = [1,2,4,5,7,8,9,0]
print(lst[:4])
# Output: [1,2,4,5]
```

```
lst = [2,3,4,5,6,8,9,0]
print(lst[-4:])
# Output: [6,8,9,0]
```

```
lst = [1,2,2,6,2,4,5,7,8,9]
print(lst[::2])
# Output: every second element
```

```
t = (0,7,5,9,3,4,1)
print(t[2:6])
# Output: slice from index 2 to 5
```

```
t = (1,2,3,4,5,6,7)
print(t[::-1])
# Output: reversed tuple
```

```
t1 = (3,1,4)
t2 = (2,6,5)
combined = tuple(sorted(t1 + t2))
print(combined)
# Output: sorted combined tuple
```

```
names = ['paq','ih','eqw','bc']
print(sorted(names))
# Output: ['bc','eqw','ih','paq']
```

```
a = ['ac','pq']
b = ['vo','cy','id','bu']
merged = sorted(a + b)
print(merged)
# Output: merged sorted list
```

```
lst = list(range(10))
half = len(lst)//2
h1 = lst[:half][::-1]
h2 = lst[half:][::-1]
print(h1)
print(h2)
# Output: reversed halves
```
