1. Marks of 10 students — mean, median, variance, std dev.

```python
import numpy as np
marks = np.array([78,85,62,90,56,73,88,95,67,80])
print('Marks:', marks.tolist())
print('Mean:', marks.mean())
print('Median:', np.median(marks))
print('Variance:', marks.var())
print('Std Dev:', marks.std())
```
