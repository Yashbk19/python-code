1. employees.csv — filter Age>30; group by Department

```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Alice","Bob","Charlie","David","Eva","Frank"],
    "Age": [28,34,45,23,38,41],
    "Department": ["HR","Sales","IT","Sales","HR","IT"],
    "Salary": [50000,60000,75000,45000,68000,72000]
})
print("Employees older than 30:")
print(df[df["Age"]>30].to_string(index=False))
print("\nSalary by Department:")
print(df.groupby("Department")["Salary"].agg(["mean","min","max"]).to_string())
```

3. student_marks.csv — filter Maths>80; totals & averages

```python
import pandas as pd
df = pd.DataFrame({
    "StudentID": [1,2,3,4,5],
    "Name": ["A","B","C","D","E"],
    "Maths": [85,76,90,66,92],
    "Science": [78,89,88,74,81],
    "English": [82,71,79,68,77]
})
print("Students with Maths>80:")
print(df[df["Maths"]>80][["StudentID","Name","Maths"]].to_string(index=False))
df["Total"] = df[["Maths","Science","English"]].sum(axis=1)
df["Avg"] = df[["Maths","Science","English"]].mean(axis=1)
print("\nTotal & Average per student:")
print(df[["Name","Total","Avg"]].to_string(index=False))
```

11. hospital.csv — filter admitted in last month; group by Department

```python
import pandas as pd
df = pd.DataFrame({
    "PatientID": range(1,11),
    "Name": ["P"+str(i) for i in range(1,11)],
    "Department": ["Cardio","Ortho","Cardio","ENT","Ortho","ENT","Cardio","ENT","Ortho","Cardio"],
    "AdmissionDate": pd.date_range("2025-09-01", periods=10, freq='3D')
})
recent_cutoff = df["AdmissionDate"].max() - pd.Timedelta(days=30)
print("Patients admitted in last 30 days:")
print(df[df["AdmissionDate"] > recent_cutoff].to_string(index=False))
print("\nPatients per Department:")
print(df.groupby("Department")["PatientID"].count().to_string())
```

13. flights.csv — filter Delay>30; group by Airline
