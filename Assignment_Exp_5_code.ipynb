```
year = 2020
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print('Leap year')
else:
    print('Not leap year')
# 2020 -> Leap year
```

```
n = 0
if n > 0:
    print('Positive')
elif n == 0:
    print('Zero')
else:
    print('Negative')
# Example: Zero
```

```
n = 7
print('Even' if n%2==0 else 'Odd')
# 7 -> Odd
```

```
ch = 'b'
if ch.lower() in 'aeiou':
    print('Vowel')
else:
    print('Consonant')
# b -> Consonant
```

```
n = 55
if n % 5 == 0 and n % 11 == 0:
    print('Divisible by both 5 and 11')
else:
    print('Not divisible by both')
# 55 -> Divisible by both
```

```
a, b, c = 3,4,5
if a + b > c and b + c > a and a + c > b:
    print('Valid triangle')
else:
    print('Invalid triangle')
# 3,4,5 -> Valid
```

```
age = 19
print('Eligible' if age >= 18 else 'Not eligible')
# 19 -> Eligible
```

```
marks = 88
if marks >= 90:
    print('A')
elif marks >= 75:
    print('B')
elif marks >= 50:
    print('C')
else:
    print('F')
# 88 -> B
```

```
n = 21
if n % 3 == 0 or n % 7 == 0:
    print('Multiple of 3 or 7')
else:
    print('Not')
# 21 -> Multiple
```

```
year = 2000
print('Century' if year % 100 == 0 else 'Not century')
# 2000 -> Century
```

```
temp = 36
if temp > 35:
    print('Hot')
elif temp >= 20:
    print('Warm')
else:
    print('Cold')
# 36 -> Hot
```
